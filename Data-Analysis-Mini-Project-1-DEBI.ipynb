```python
def clean_dataframe(df):
    # 1. إزالة التكرار
    df = df.drop_duplicates().copy()

    # 2. معالجة القيم الناقصة حسب نوع العمود
    for col in df.columns:
        if df[col].dtype in ['int64', 'float64']:
            df.loc[:, col] = df[col].fillna(0)
        else:
            df.loc[:, col] = df[col].fillna("Unknown")

    # 3. إزالة الأوتلايرز باستخدام IQR
    def remove_outliers_iqr(data, col):
        Q1 = data[col].quantile(0.25)
        Q3 = data[col].quantile(0.75)
        IQR = Q3 - Q1
        lower_bound = Q1 - 1.5 * IQR
        upper_bound = Q3 + 1.5 * IQR
        return data[(data[col] >= lower_bound) & (data[col] <= upper_bound)]

    cols_to_clean = ['Sales', 'Profit', 'Discount', 'Quantity']
    for col in cols_to_clean:
        df = remove_outliers_iqr(df, col)

    # 4. تحويل الأعمدة للتواريخ
    df['Order Date'] = pd.to_datetime(df['Order Date'], errors='coerce')
    df['Ship Date'] = pd.to_datetime(df['Ship Date'], errors='coerce')

    # 5. تحويل المعرفات لأعداد صحيحة
    df['Row ID'] = df['Row ID'].astype('int64')
    df['Postal Code'] = df['Postal Code'].astype('Int64')  # يدعم القيم الفاضية

    # 6. مراجعة الأعمدة الرقمية
    numeric_cols = ['Sales', 'Quantity', 'Discount', 'Profit']
    for col in numeric_cols:
        df[col] = pd.to_numeric(df[col], errors='coerce')

    # 7. إزالة الصفوف اللي ناقص فيها تواريخ
    df = df.dropna(subset=['Order Date', 'Ship Date'])


    print(" Data Cleaning Completed")
    print("Number of rows after cleaning:", len(df))
    print("\nMissing values per column:")
    print(df.isnull().sum())
    print("\nData types after cleaning:")
    print(df.dtypes)

    return df
```

```python
import pandas as pd

def calculate_kpis(df):
    """
    Calculate the main KPIs of the dataset.
    """

    kpis = {
        'Total Sales': df['Sales'].sum(),
        'Total Profit': df['Profit'].sum(),
        'Total Quantity': df['Quantity'].sum(),
        'Average Discount': df['Discount'].mean(),
        'Average Profit Margin': df['Proft Margin'].mean(), # Corrected column name
        'Average Shipping Duration': df['shipping_time'].mean().total_seconds() / (24 * 3600), # Corrected column name and converted to days
        'Number of Orders': df['Order ID'].nunique(),
        'Number of Customers': df['Customer ID'].nunique()
    }

    return kpis

kpis = calculate_kpis(df)

for key, value in kpis.items():
    print(f"{key}: {value:,.2f}")
```

```python
def correlation_analysis(df, columns):
    """
    Calculate correlation between selected numerical columns.
    """

    return df[columns].corr()
```

```python
correlation_result = correlation_analysis(
    df,
    [
        'Sales',
        'Quantity',
        'Discount',
        'Profit',
        'Proft Margin',
        'shipping_time' # Corrected column name
    ]
)

display(correlation_result)
```
